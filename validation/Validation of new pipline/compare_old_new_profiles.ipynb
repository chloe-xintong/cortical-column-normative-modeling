```python
old_files = sorted(
    OLD_PROFILE_DIR.glob(
        f"{SUBJECT}_*_cols_adc.csv"
    )
)

new_files = sorted(
    NEW_PROFILE_DIR.glob(
        f"{SUBJECT}_*_md.csv"
    )
)

print("Old regions:", len(old_files))
print("New regions:", len(new_files))
def load_profile_csv(path):
    df = pd.read_csv(path)

    values = (
        df.select_dtypes(
            include=[np.number]
        )
        .to_numpy()
    )

    if values.shape[1] > 21:
        values = values[:, -21:]

    return values
```

```python
import numpy as np
import pandas as pd
from scipy.io import loadmat


def load_region_coordinates(path):
    data = loadmat(path)

    keys = [
        key for key in data.keys()
        if not key.startswith("__")
    ]

    if len(keys) != 1:
        raise ValueError(
            f"Expected one data variable in {path}, found: {keys}"
        )

    coords = np.asarray(data[keys[0]])

    if coords.shape[0] != 4:
        raise ValueError(
            f"Expected shape (4, N), got {coords.shape}"
        )

    n_points = coords.shape[1]

    if n_points % 21 != 0:
        raise ValueError(
            f"Number of points is not divisible by 21: {n_points}"
        )

    # Remove homogeneous coordinate and reshape:
    # (4, N*21) -> (N, 21, 3)
    coords = coords[:3, :].T
    coords = coords.reshape(-1, 21, 3)

    return coords


rows = []

old_files = sorted(OLD_REGION_DIR.glob("*.mat"))

for old_path in old_files:
    region = old_path.stem
    new_path = NEW_REGION_DIR / f"{region}.mat"

    if not new_path.exists():
        continue

    old_coords = load_region_coordinates(old_path)
    new_coords = load_region_coordinates(new_path)

    # Mean sampling coordinate at each cortical depth
    old_mean = np.mean(old_coords, axis=0)
    new_mean = np.mean(new_coords, axis=0)

    # Euclidean OLD-NEW distance at each depth
    depth_distance = np.linalg.norm(
        new_mean - old_mean,
        axis=1,
    )

    rows.append({
        "region": region,
        "old_n_trajectories": old_coords.shape[0],
        "new_n_trajectories": new_coords.shape[0],
        "mean_spatial_distance_vox": np.mean(depth_distance),
        "max_spatial_distance_vox": np.max(depth_distance),
        "white_distance_vox": depth_distance[0],
        "mid_distance_vox": depth_distance[10],
        "pial_distance_vox": depth_distance[20],
    })


geometry_df = pd.DataFrame(rows)

geometry_df = geometry_df.sort_values(
    "mean_spatial_distance_vox",
    ascending=False,
).reset_index(drop=True)

geometry_df
```

```python
from scipy.spatial import cKDTree


def symmetric_nearest_neighbor_distance(old_points, new_points):
    old_tree = cKDTree(old_points)
    new_tree = cKDTree(new_points)

    old_to_new, _ = new_tree.query(old_points, k=1)
    new_to_old, _ = old_tree.query(new_points, k=1)

    return (
        np.mean(old_to_new) + np.mean(new_to_old)
    ) / 2


rows = []

for old_path in sorted(OLD_REGION_DIR.glob("*.mat")):
    region = old_path.stem
    new_path = NEW_REGION_DIR / f"{region}.mat"

    if not new_path.exists():
        continue

    old_coords = load_region_coordinates(old_path)
    new_coords = load_region_coordinates(new_path)

    depth_distances = []

    for depth in range(21):
        old_points = old_coords[:, depth, :]
        new_points = new_coords[:, depth, :]

        distance = symmetric_nearest_neighbor_distance(
            old_points,
            new_points,
        )

        depth_distances.append(distance)

    rows.append({
        "region": region,
        "mean_nn_distance_vox": np.mean(depth_distances),
        "max_nn_distance_vox": np.max(depth_distances),
        "white_nn_distance_vox": depth_distances[0],
        "mid_nn_distance_vox": depth_distances[10],
        "pial_nn_distance_vox": depth_distances[20],
    })


distribution_geometry_df = pd.DataFrame(rows)

distribution_geometry_df = distribution_geometry_df.sort_values(
    "mean_nn_distance_vox",
    ascending=False,
).reset_index(drop=True)

distribution_geometry_df.head(20)
```

```python
import numpy as np
import pandas as pd
import nibabel as nib

from scipy.io import loadmat
from scipy.ndimage import map_coordinates


def load_region_coordinates(path):
    data = loadmat(path)

    keys = [
        key for key in data.keys()
        if not key.startswith("__")
    ]

    if len(keys) != 1:
        raise ValueError(
            f"Expected one data variable in {path}, found: {keys}"
        )

    coords = np.asarray(data[keys[0]])

    if coords.shape[0] != 4:
        raise ValueError(
            f"Expected shape (4, N), got {coords.shape}"
        )

    if coords.shape[1] % 21 != 0:
        raise ValueError(
            f"Number of points is not divisible by 21: "
            f"{coords.shape[1]}"
        )

    coords = coords[:3, :].T
    coords = coords.reshape(-1, 21, 3)

    return coords


def sample_volume(volume, coords):
    flat_coords = coords.reshape(-1, 3)

    sample_coords = [
        flat_coords[:, 0],
        flat_coords[:, 1],
        flat_coords[:, 2],
    ]

    sampled = map_coordinates(
        volume,
        sample_coords,
        order=1,
        mode="nearest",
    )

    return sampled.reshape(-1, 21)
```

```python
import numpy as np
import pandas as pd
import nibabel as nib

from pathlib import Path
from scipy.io import loadmat
from scipy.ndimage import map_coordinates


SUBJECT = "D0007"
DEPTH_SAMPLES = 21

TEST_ROOT = Path(
    "/mnt/newStor/paros/paros_WORK/xintong/test"
)

OLD_ROOT = (
    TEST_ROOT
    / "existing_pipeline"
    / "output"
    / SUBJECT
)

NEW_ROOT = (
    TEST_ROOT
    / "new_pipeline"
    / SUBJECT
)

OLD_COLUMN_DIR = OLD_ROOT / "columns"

NEW_TRAJECTORY_DIR = (
    NEW_ROOT
    / "trajectories"
)

NEW_ADC = (
    NEW_ROOT
    / "diffusion"
    / f"{SUBJECT}_adc.nii.gz"
)

OLD_REG = (
    OLD_ROOT
    / "DWI2T1_dti.dat"
)

NEW_REG = (
    NEW_ROOT
    / "registration"
    / "DWI2T1_dti_masked.dat"
)


def load_transform_matrix(path):
    rows = []

    with open(path, "r") as f:
        for line in f:
            values = line.strip().split()

            if len(values) != 4:
                continue

            try:
                row = [
                    float(value)
                    for value in values
                ]
            except ValueError:
                continue

            rows.append(row)

    if len(rows) < 4:
        raise ValueError(
            f"Could not find 4x4 matrix in {path}"
        )

    return np.asarray(
        rows[:4],
        dtype=float,
    )


def vox2ras_tkreg(shape, voxel_size):
    nx, ny, nz = [
        float(value)
        for value in shape
    ]

    dx, dy, dz = [
        float(value)
        for value in voxel_size
    ]

    matrix = np.zeros(
        (4, 4),
        dtype=float,
    )

    matrix[0, 0] = -dx
    matrix[0, 3] = dx * nx / 2.0

    matrix[1, 2] = dz
    matrix[1, 3] = -dz * nz / 2.0

    matrix[2, 1] = -dy
    matrix[2, 3] = dy * ny / 2.0

    matrix[3, 3] = 1.0

    return matrix


def vox2ras_0to1(matrix):
    q = np.zeros(
        (4, 4),
        dtype=float,
    )

    q[:3, 3] = 1.0

    return np.linalg.inv(
        np.linalg.inv(matrix) + q
    )


def load_t1_trajectories(path):
    data = loadmat(path)

    keys = [
        key
        for key in data.keys()
        if not key.startswith("__")
    ]

    if len(keys) != 1:
        print(
            f"{path.name}: variables = {keys}"
        )

    key = keys[0]

    values = np.asarray(
        data[key],
        dtype=float,
    )

    # Common flattened format: 3 x (N*21)
    if (
        values.ndim == 2
        and values.shape[0] == 3
    ):
        points = values.T

    # Homogeneous format: 4 x (N*21)
    elif (
        values.ndim == 2
        and values.shape[0] == 4
    ):
        points = values[:3, :].T

    # Already transposed
    elif (
        values.ndim == 2
        and values.shape[1] in (3, 4)
    ):
        points = values[:, :3]

    # Already N x 21 x 3
    elif (
        values.ndim == 3
        and values.shape[-1] == 3
    ):
        return values

    else:
        raise ValueError(
            f"Unexpected shape in {path}: "
            f"{values.shape}"
        )

    if (
        points.shape[0]
        % DEPTH_SAMPLES
        != 0
    ):
        raise ValueError(
            f"{path}: point count "
            f"{points.shape[0]} "
            "is not divisible by 21"
        )

    return points.reshape(
        -1,
        DEPTH_SAMPLES,
        3,
    )


def transform_to_dwi(
    trajectories,
    transform,
    t_mov,
):
    points = trajectories.reshape(
        -1,
        3,
    )

    homogeneous = np.ones(
        (4, points.shape[0]),
        dtype=float,
    )

    homogeneous[:3, :] = points.T

    transformed = (
        np.linalg.inv(t_mov)
        @ transform
        @ homogeneous
    )

    return (
        transformed[:3, :]
        .T
        .reshape(
            -1,
            DEPTH_SAMPLES,
            3,
        )
    )


def sample_volume(
    volume,
    coords,
):
    # Coordinates are stored using 1-based
    # voxel indexing in this pipeline.
    sample_coords = (
        coords.reshape(-1, 3).T
        - 1.0
    )

    sampled = map_coordinates(
        volume,
        sample_coords,
        order=1,
        mode="nearest",
    )

    return sampled.reshape(
        -1,
        DEPTH_SAMPLES,
    )


def compare_profiles(
    profile_a,
    profile_b,
):
    profile_a = np.asarray(
        profile_a,
        dtype=float,
    )

    profile_b = np.asarray(
        profile_b,
        dtype=float,
    )

    diff = profile_b - profile_a

    return {
        "correlation": np.corrcoef(
            profile_a,
            profile_b,
        )[0, 1],
        "mae": np.mean(
            np.abs(diff)
        ),
        "rmse": np.sqrt(
            np.mean(diff ** 2)
        ),
        "mean_difference": np.mean(
            diff
        ),
    }


# Load NEW ADC.
new_adc_img = nib.load(
    str(NEW_ADC)
)

new_volume = (
    new_adc_img.get_fdata()
)

shape = (
    new_adc_img.shape[:3]
)

voxel_size = (
    new_adc_img
    .header
    .get_zooms()[:3]
)

t_mov = vox2ras_0to1(
    vox2ras_tkreg(
        shape,
        voxel_size,
    )
)

old_transform = (
    load_transform_matrix(
        OLD_REG
    )
)

new_transform = (
    load_transform_matrix(
        NEW_REG
    )
)


# Registration-only experiment
registration_old_profiles = []
registration_new_profiles = []

registration_rows = []

for hemi in ["lh", "rh"]:

    old_t1 = (
        load_t1_trajectories(
            OLD_COLUMN_DIR
            / f"{SUBJECT}_column_{hemi}.mat"
        )
    )

    coords_old_reg = (
        transform_to_dwi(
            old_t1,
            old_transform,
            t_mov,
        )
    )

    coords_new_reg = (
        transform_to_dwi(
            old_t1,
            new_transform,
            t_mov,
        )
    )

    sampled_old_reg = sample_volume(
        new_volume,
        coords_old_reg,
    )

    sampled_new_reg = sample_volume(
        new_volume,
        coords_new_reg,
    )

    profile_old_reg = np.mean(
        sampled_old_reg,
        axis=0,
    )

    profile_new_reg = np.mean(
        sampled_new_reg,
        axis=0,
    )

    metrics = compare_profiles(
        profile_old_reg,
        profile_new_reg,
    )

    registration_rows.append({
        "hemisphere": hemi,
        "n_trajectories":
            old_t1.shape[0],
        **metrics,
    })

    registration_old_profiles.append(
        profile_old_reg
    )

    registration_new_profiles.append(
        profile_new_reg
    )


registration_old_all = np.concatenate(
    registration_old_profiles
)

registration_new_all = np.concatenate(
    registration_new_profiles
)

registration_overall = {
    "experiment":
        "Registration only",
    **compare_profiles(
        registration_old_all,
        registration_new_all,
    ),
}


# T1 trajectory / FreeSurfer-only experiment
trajectory_old_profiles = []
trajectory_new_profiles = []

trajectory_rows = []

for hemi in ["lh", "rh"]:

    old_t1 = (
        load_t1_trajectories(
            OLD_COLUMN_DIR
            / f"{SUBJECT}_column_{hemi}.mat"
        )
    )

    new_t1 = (
        load_t1_trajectories(
            NEW_TRAJECTORY_DIR
            / f"{SUBJECT}_trajectory_{hemi}.mat"
        )
    )

    # Same registration for both.
    old_coords = (
        transform_to_dwi(
            old_t1,
            new_transform,
            t_mov,
        )
    )

    new_coords = (
        transform_to_dwi(
            new_t1,
            new_transform,
            t_mov,
        )
    )

    old_sampled = sample_volume(
        new_volume,
        old_coords,
    )

    new_sampled = sample_volume(
        new_volume,
        new_coords,
    )

    old_profile = np.mean(
        old_sampled,
        axis=0,
    )

    new_profile = np.mean(
        new_sampled,
        axis=0,
    )

    metrics = compare_profiles(
        old_profile,
        new_profile,
    )

    trajectory_rows.append({
        "hemisphere": hemi,
        "old_n_trajectories":
            old_t1.shape[0],
        "new_n_trajectories":
            new_t1.shape[0],
        **metrics,
    })

    trajectory_old_profiles.append(
        old_profile
    )

    trajectory_new_profiles.append(
        new_profile
    )


trajectory_old_all = np.concatenate(
    trajectory_old_profiles
)

trajectory_new_all = np.concatenate(
    trajectory_new_profiles
)

trajectory_overall = {
    "experiment":
        "T1 trajectory / FreeSurfer only",
    **compare_profiles(
        trajectory_old_all,
        trajectory_new_all,
    ),
}


# Previously measured results
previous_results = [
    {
        "experiment":
            "Full end-to-end",
        "correlation": 0.942154,
        "mae": 0.000059,
        "rmse": 0.000087,
        "mean_difference":
            0.000031,
    },
    {
        "experiment":
            "Image only",
        "correlation": 0.995122,
        "mae": 0.000007,
        "rmse": 0.000023,
        "mean_difference":
            -1.190009e-07,
    },
    {
        "experiment":
            "Full geometry",
        "correlation": 0.942715,
        "mae": 0.000055,
        "rmse": 0.000076,
        "mean_difference":
            0.000011,
    },
]


summary_df = pd.DataFrame(
    previous_results
    + [
        registration_overall,
        trajectory_overall,
    ]
)

full_mae = (
    summary_df.loc[
        summary_df["experiment"]
        == "Full end-to-end",
        "mae",
    ]
    .iloc[0]
)

summary_df[
    "mae_relative_to_end_to_end"
] = (
    summary_df["mae"]
    / full_mae
)

print("\nRegistration-only by hemisphere:")
display(
    pd.DataFrame(
        registration_rows
    )
)

print(
    "\nT1 trajectory / FreeSurfer-only "
    "by hemisphere:"
)
display(
    pd.DataFrame(
        trajectory_rows
    )
)

print("\nFinal comparison:")
display(summary_df)
```

```python
from pathlib import Path

import numpy as np
import pandas as pd
import nibabel as nib

from scipy.io import loadmat
from scipy.ndimage import map_coordinates


# Inputs
SUBJECT = "D0007"
N_DEPTH = 21

TEST_ROOT = Path(
    "/mnt/newStor/paros/paros_WORK/xintong/test"
)

OLD_ROOT = (
    TEST_ROOT
    / "existing_pipeline"
    / "output"
    / SUBJECT
)

NEW_ROOT = (
    TEST_ROOT
    / "new_pipeline"
    / SUBJECT
)

OLD_REGION_DIR = (
    OLD_ROOT
    / "columns"
    / "label_coord_1mm"
)

NEW_REGION_DIR = (
    NEW_ROOT
    / "regions"
)

NEW_ADC = (
    NEW_ROOT
    / "diffusion"
    / f"{SUBJECT}_adc.nii.gz"
)

OLD_REG = (
    OLD_ROOT
    / "DWI2T1_dti.dat"
)

NEW_REG = (
    NEW_ROOT
    / "registration"
    / "DWI2T1_dti_masked.dat"
)


# Load bbregister matrix
def load_transform_matrix(path):
    rows = []

    with open(path, "r") as file:
        for line in file:
            values = line.strip().split()

            if len(values) != 4:
                continue

            try:
                row = [
                    float(value)
                    for value in values
                ]
            except ValueError:
                continue

            rows.append(row)

    if len(rows) < 4:
        raise ValueError(
            f"Could not find 4x4 matrix in {path}"
        )

    return np.asarray(
        rows[:4],
        dtype=float,
    )


# Reproduce FreeSurfer tkregister voxel geometry
def vox2ras_tkreg(shape, voxel_size):
    nx, ny, nz = [
        float(value)
        for value in shape
    ]

    dx, dy, dz = [
        float(value)
        for value in voxel_size
    ]

    matrix = np.zeros(
        (4, 4),
        dtype=float,
    )

    matrix[0, 0] = -dx
    matrix[0, 3] = dx * nx / 2.0

    matrix[1, 2] = dz
    matrix[1, 3] = -dz * nz / 2.0

    matrix[2, 1] = -dy
    matrix[2, 3] = dy * ny / 2.0

    matrix[3, 3] = 1.0

    return matrix


def vox2ras_0to1(matrix):
    q = np.zeros(
        (4, 4),
        dtype=float,
    )

    q[:3, 3] = 1.0

    return np.linalg.inv(
        np.linalg.inv(matrix) + q
    )


# Load one region's DWI-space trajectory coordinates
def load_region_coordinates(path):
    data = loadmat(path)

    keys = [
        key
        for key in data.keys()
        if not key.startswith("__")
    ]

    coordinate_keys = [
        key
        for key in keys
        if (
            "dwi" in key.lower()
            or "coord" in key.lower()
        )
    ]

    if coordinate_keys:
        key = coordinate_keys[0]

    elif len(keys) == 1:
        key = keys[0]

    else:
        raise ValueError(
            f"Could not identify coordinate variable "
            f"in {path}: {keys}"
        )

    coords = np.asarray(
        data[key],
        dtype=float,
    )

    if coords.ndim != 2:
        raise ValueError(
            f"Unexpected shape in {path}: "
            f"{coords.shape}"
        )

    if coords.shape[0] in (3, 4):
        coords = coords[:3, :].T

    elif coords.shape[1] in (3, 4):
        coords = coords[:, :3]

    else:
        raise ValueError(
            f"Could not interpret coordinates "
            f"in {path}: {coords.shape}"
        )

    if coords.shape[0] % N_DEPTH != 0:
        raise ValueError(
            f"{path}: point count "
            f"{coords.shape[0]} "
            f"is not divisible by {N_DEPTH}"
        )

    return coords.reshape(
        -1,
        N_DEPTH,
        3,
    )


# Convert DWI voxel coordinates back to T1 tkregister coordinates
def dwi_to_t1(
    dwi_coords,
    transform,
    t_mov,
):
    points = dwi_coords.reshape(
        -1,
        3,
    )

    homogeneous = np.ones(
        (4, points.shape[0]),
        dtype=float,
    )

    homogeneous[:3, :] = points.T

    t1_points = (
        np.linalg.inv(transform)
        @ t_mov
        @ homogeneous
    )

    return (
        t1_points[:3, :]
        .T
        .reshape(
            -1,
            N_DEPTH,
            3,
        )
    )


# Transform T1 coordinates into DWI voxel coordinates
def t1_to_dwi(
    t1_coords,
    transform,
    t_mov,
):
    points = t1_coords.reshape(
        -1,
        3,
    )

    homogeneous = np.ones(
        (4, points.shape[0]),
        dtype=float,
    )

    homogeneous[:3, :] = points.T

    dwi_points = (
        np.linalg.inv(t_mov)
        @ transform
        @ homogeneous
    )

    return (
        dwi_points[:3, :]
        .T
        .reshape(
            -1,
            N_DEPTH,
            3,
        )
    )


# Sample ADC using the same 1-based voxel convention
def sample_volume(
    volume,
    coords,
):
    sample_coords = (
        coords.reshape(-1, 3).T
        - 1.0
    )

    sampled = map_coordinates(
        volume,
        sample_coords,
        order=1,
        mode="nearest",
    )

    return sampled.reshape(
        -1,
        N_DEPTH,
    )


def compare_profiles(
    old_profile,
    new_profile,
):
    old_profile = np.asarray(
        old_profile,
        dtype=float,
    )

    new_profile = np.asarray(
        new_profile,
        dtype=float,
    )

    diff = (
        new_profile
        - old_profile
    )

    return {
        "correlation": np.corrcoef(
            old_profile,
            new_profile,
        )[0, 1],

        "mae": np.mean(
            np.abs(diff)
        ),

        "rmse": np.sqrt(
            np.mean(diff ** 2)
        ),

        "mean_difference": np.mean(
            diff
        ),
    }


# Load NEW ADC and geometry
new_adc_img = nib.load(
    str(NEW_ADC)
)

new_volume = (
    new_adc_img.get_fdata()
)

shape = (
    new_adc_img.shape[:3]
)

voxel_size = (
    new_adc_img
    .header
    .get_zooms()[:3]
)

t_mov = vox2ras_0to1(
    vox2ras_tkreg(
        shape,
        voxel_size,
    )
)

old_transform = (
    load_transform_matrix(
        OLD_REG
    )
)

new_transform = (
    load_transform_matrix(
        NEW_REG
    )
)


registration_rows = []
freesurfer_rows = []

registration_old_profiles = []
registration_new_profiles = []

freesurfer_old_profiles = []
freesurfer_new_profiles = []


# Run all 68 cortical regions
for old_path in sorted(
    OLD_REGION_DIR.glob("*.mat")
):
    region = old_path.stem

    new_path = (
        NEW_REGION_DIR
        / f"{region}.mat"
    )

    if not new_path.exists():
        print(
            f"Skipping missing NEW region: "
            f"{region}"
        )
        continue

    # Existing final DWI coordinates
    old_dwi_coords = (
        load_region_coordinates(
            old_path
        )
    )

    new_dwi_coords = (
        load_region_coordinates(
            new_path
        )
    )

    # Recover each pipeline's T1-space coordinates.
    old_t1_coords = dwi_to_t1(
        old_dwi_coords,
        old_transform,
        t_mov,
    )

    new_t1_coords = dwi_to_t1(
        new_dwi_coords,
        new_transform,
        t_mov,
    )


    # -------------------------------------------------
    # Experiment 1:
    # Registration only
    #
    # Same OLD T1 coordinates
    # Same NEW ADC
    # Only OLD vs NEW registration changes
    # -------------------------------------------------

    reg_old_coords = t1_to_dwi(
        old_t1_coords,
        old_transform,
        t_mov,
    )

    reg_new_coords = t1_to_dwi(
        old_t1_coords,
        new_transform,
        t_mov,
    )

    reg_old_values = sample_volume(
        new_volume,
        reg_old_coords,
    )

    reg_new_values = sample_volume(
        new_volume,
        reg_new_coords,
    )

    reg_old_profile = np.mean(
        reg_old_values,
        axis=0,
    )

    reg_new_profile = np.mean(
        reg_new_values,
        axis=0,
    )

    reg_metrics = compare_profiles(
        reg_old_profile,
        reg_new_profile,
    )

    registration_rows.append({
        "region": region,
        "n_trajectories":
            old_t1_coords.shape[0],
        **reg_metrics,
    })

    registration_old_profiles.append(
        reg_old_profile
    )

    registration_new_profiles.append(
        reg_new_profile
    )


    # -------------------------------------------------
    # Experiment 2:
    # T1 trajectory / FreeSurfer only
    #
    # Same NEW registration
    # Same NEW ADC
    # OLD vs NEW T1 geometry changes
    # -------------------------------------------------

    fs_old_coords = t1_to_dwi(
        old_t1_coords,
        new_transform,
        t_mov,
    )

    fs_new_coords = t1_to_dwi(
        new_t1_coords,
        new_transform,
        t_mov,
    )

    fs_old_values = sample_volume(
        new_volume,
        fs_old_coords,
    )

    fs_new_values = sample_volume(
        new_volume,
        fs_new_coords,
    )

    fs_old_profile = np.mean(
        fs_old_values,
        axis=0,
    )

    fs_new_profile = np.mean(
        fs_new_values,
        axis=0,
    )

    fs_metrics = compare_profiles(
        fs_old_profile,
        fs_new_profile,
    )

    freesurfer_rows.append({
        "region": region,
        "old_n_trajectories":
            old_t1_coords.shape[0],
        "new_n_trajectories":
            new_t1_coords.shape[0],
        **fs_metrics,
    })

    freesurfer_old_profiles.append(
        fs_old_profile
    )

    freesurfer_new_profiles.append(
        fs_new_profile
    )


# Region-level tables
registration_region_df = pd.DataFrame(
    registration_rows
)

freesurfer_region_df = pd.DataFrame(
    freesurfer_rows
)


# Overall registration-only result
registration_old_all = np.concatenate(
    registration_old_profiles
)

registration_new_all = np.concatenate(
    registration_new_profiles
)

registration_overall = {
    "experiment":
        "Registration only (68 regions)",

    **compare_profiles(
        registration_old_all,
        registration_new_all,
    ),
}


# Overall FreeSurfer / trajectory-only result
freesurfer_old_all = np.concatenate(
    freesurfer_old_profiles
)

freesurfer_new_all = np.concatenate(
    freesurfer_new_profiles
)

freesurfer_overall = {
    "experiment":
        "T1 trajectory / FreeSurfer only (68 regions)",

    **compare_profiles(
        freesurfer_old_all,
        freesurfer_new_all,
    ),
}


# Previously measured controlled experiments
previous_results = [
    {
        "experiment":
            "Full end-to-end",

        "correlation":
            0.942154,

        "mae":
            0.000059,

        "rmse":
            0.000087,

        "mean_difference":
            0.000031,
    },

    {
        "experiment":
            "Image only",

        "correlation":
            0.995122,

        "mae":
            0.000007,

        "rmse":
            0.000023,

        "mean_difference":
            -1.190009e-07,
    },

    {
        "experiment":
            "Full geometry",

        "correlation":
            0.942715,

        "mae":
            0.000055,

        "rmse":
            0.000076,

        "mean_difference":
            0.000011,
    },
]


summary_df = pd.DataFrame(
    previous_results
    + [
        registration_overall,
        freesurfer_overall,
    ]
)


full_mae = (
    summary_df.loc[
        summary_df["experiment"]
        == "Full end-to-end",
        "mae",
    ]
    .iloc[0]
)

summary_df[
    "mae_relative_to_end_to_end"
] = (
    summary_df["mae"]
    / full_mae
)


print(
    "\nRegistration-only: "
    "regions with largest MAE"
)

display(
    registration_region_df
    .sort_values(
        "mae",
        ascending=False,
    )
    .head(20)
)


print(
    "\nT1 trajectory / FreeSurfer-only: "
    "regions with largest MAE"
)

display(
    freesurfer_region_df
    .sort_values(
        "mae",
        ascending=False,
    )
    .head(20)
)


print(
    "\nFinal 68-region comparison:"
)

display(
    summary_df
)
```

```python
import numpy as np
from pathlib import Path


def load_matrix(path):
    lines = Path(path).read_text().splitlines()

    for start in range(len(lines) - 3):
        rows = []

        for line in lines[start:start + 4]:
            parts = line.strip().split()

            if len(parts) < 4:
                break

            try:
                rows.append([float(x) for x in parts[:4]])
            except ValueError:
                break

        if len(rows) == 4:
            return np.asarray(rows)

    raise ValueError("4x4 matrix not found")


old_transform = load_matrix(
    "/mnt/newStor/paros/paros_WORK/xintong/test/existing_pipeline/output/"
    "D0007/DWI2T1_dti.dat"
)

new_transform = load_matrix(
    "/mnt/newStor/paros/paros_WORK/xintong/test/new_pipeline/D0007/registration/"
    "DWI2T1_dti_sum_masked.dat"
)

difference = new_transform - old_transform

print("OLD:")
print(old_transform)

print("\nNEW:")
print(new_transform)

print("\nNEW - OLD:")
print(difference)

print(
    "\nMax absolute element difference:",
    np.max(np.abs(difference)),
)

translation_difference = (
    new_transform[:3, 3]
    - old_transform[:3, 3]
)

print(
    "Translation difference:",
    translation_difference,
)

print(
    "Translation difference magnitude:",
    np.linalg.norm(translation_difference),
)
```
